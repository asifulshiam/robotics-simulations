```python
import numpy as np
import matplotlib.pyplot as plt

# For animation
import matplotlib.animation as animation
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

# **Question 2**

```python
# Question 2 dt = 0.1
# variables
v = 1
ti = 0
tf = 10
dt = 0.1

time = np.arange(ti, tf+dt, dt)
#print(time)
n = len(time)

w = np.zeros(n)
#print(w)

# values of w in the list
for i in range(n):
  if 0.5 <= time[i] <=1.5:
    w[i] = 3
  elif 2 <= time[i] <= 3:
    w[i] = -3
  elif 4 <= time[i] <= 5:
    w[i] = -3
  elif 6 <= time[i] <= 7:
    w[i] = 3
  elif 8 <= time[i] <= 9:
    w[i] = -3
  else:
    w[i] = 0

#for i in range(n):
  #print(w[i])

# initializing list for x and y to get values by Euler's Equation
x = np.zeros(len(time))
y = np.zeros(len(time))
theta = np.zeros(len(time))

# setting initial conditions
x[0] = 0
y[0] = 0
theta[0] = 1

# Using Euler's Integration
# i means we working with loops
for i in range(0,n-1):
  x[i+1] = x[i] + v*np.cos(theta[i])*dt
  y[i+1] = y[i] + v*np.sin(theta[i])*dt
  theta[i+1] = theta[i] + w[i]*dt

# Part 2a
plt.figure(figsize=(12,10))

# xy plot
plt.subplot(2,2,1)
plt.plot(x, y)
plt.title("x vs y")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

# xt plot
plt.subplot(2,2,2)
plt.plot(time, x)
plt.title("x vs t")
plt.xlabel("t")
plt.ylabel("x")
plt.grid()

# yt plot
plt.subplot(2,2,3)
plt.plot(time, y)
plt.title("y vs t")
plt.xlabel("t")
plt.ylabel("y")
plt.grid()

# theta,t plot
plt.subplot(2,2,4)
plt.plot(time, theta)
plt.title("theta vs t")
plt.xlabel("t")
plt.ylabel("theta")
plt.grid()

plt.show()
```

```python
# Part 2b
# Unicycle Movie

fig,ax = plt.subplots()
ax.set_xlim(np.min(x), np.max(x))
ax.set_ylim(np.min(y), np.max(y))

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('x vs y')

line, = ax.plot([],[],lw=2)

def init():
  line.set_data([],[])
  return line,

def animate(frame):
  xarr = x[:frame]
  yarr = y[:frame]
  line.set_data(xarr,yarr)
  return line,

#the frame parameter is the duration the simulation will run
ani = FuncAnimation(fig, animate, frames= range(0, len(time)), init_func= init, blit= True)
# Need to convert to HTML to view (HTML5 video)
from IPython.display import HTML
HTML(ani.to_jshtml())
```

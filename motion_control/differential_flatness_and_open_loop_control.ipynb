# Question 2)c)

```python
dt = 0.01
t = np.arange(0, T + dt, dt)

#velocity
z1_dot = alpha[1] + 2*alpha[2]*t + 3*alpha[3]*t**2
z2_dot = alpha[5] + 2*alpha[6]*t + 3*alpha[7]*t**2
#acceleration
z1_dotdot = 2*alpha[2] + 6*alpha[3]*t
z2_dotdot = 2*alpha[6] + 6*alpha[7]*t

#intialize arrays
xd = np.zeros(len(t))
yd = np.zeros(len(t))
V = np.zeros(len(t))
theta = np.zeros(len(t))

#set initial conditions
xd[0] = x10
yd[0] = y10
V[0] = v0
theta[0] = tht0
for i in range(0, len(t)-1,1):
  b2 = np.array([z1_dotdot[i], z2_dotdot[i]])
  A2 = np.array([
      [np.cos(theta[i]), -V[i]*np.sin(theta[i])],
      [np.sin(theta[i]), V[i]*np.cos(theta[i])]
  ])

  A2_inv = np.linalg.pinv(A2)
  x2 = np.dot(A2_inv,b2)

  a_t = x2[0]
  w_t = x2[1]

  #Eulers
  xd[i+1] = xd[i] + dt* V[i]*np.cos(theta[i])
  yd[i+1] = yd[i] + dt* V[i] *np.sin(theta[i])
  V[i+1] = V[i] + dt*a_t
  theta[i+1] = theta[i]+ dt*w_t

  fig = plt.figure()
  ax = fig.add_subplot(1, 1, 1)
  ax.plot(z1, z2, linestyle='', marker='o', label="Robot Trajectory")  # Trajectory
  ax.plot(xd, yd, linestyle='', marker='o', label="Robot Position")  # Robot position
  plt.title("Robot Trajectory + Robot Position")
  plt.xlabel("x")
  plt.ylabel("y")
  plt.legend()
  plt.show()
```

```python
# Final-state plot for Q2c (Δt = 0.01)
# Note: the cell above contains a plt.show() inside the integration loop,
# which renders 1500 intermediate frames when run. The outputs of that cell
# are stripped from the committed notebook to keep file size reasonable.
# This cell shows only the final overlay.

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot(z1, z2, linestyle='', marker='o', label="Desired Trajectory")
ax.plot(xd, yd, linestyle='', marker='o', label="Robot Position (open-loop, Δt = 0.01)")
plt.title("Q2c — Open-loop trajectory tracking, Δt = 0.01")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

Alternative way to avoid discretization error in euler's method

```python
v_d = np.sqrt(z1_dot**2 + z2_dot**2)
theta_d = np.arctan2(z2_dot, z1_dot)

v_d_safe = np.where(v_d < 1e-3, 1e-3, v_d)
a_d = (z1_dot*z1_dotdot + z2_dot*z2_dotdot) / v_d_safe
w_d = (z1_dot*z2_dotdot - z2_dot*z1_dotdot) / (v_d_safe**2)
```

```python
for i in range(len(t) - 1):

    xd[i+1] = xd[i] + dt * V[i] * np.cos(theta[i])
    yd[i+1] = yd[i] + dt * V[i] * np.sin(theta[i])

    # Update velocity with feed-forward acceleration:
    V[i+1] = V[i] + dt * a_d[i]
    # Clamp velocity to avoid negative values
    V[i+1] = max(0, V[i+1])
    # Update heading with feed-forward angular velocity:
    theta[i+1] = theta[i] + dt * w_d[i]
    # Wrap theta to keep it within [-pi, pi]
    theta[i+1] = (theta[i+1] + np.pi) % (2*np.pi) - np.pi

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot(z1, z2, linestyle='-', marker='o', label="Robot Trajectory")  # Planned trajectory
ax.plot(xd, yd, linestyle='-', marker='o', label="Robot Position")  # Simulated position
plt.title("Robot Trajectory + Robot Position")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
